# RangeStatus
用于描述时间范围的状态：
- `Open`：表示范围仍在进行中，终点尚未确定
- `Closed`：表示范围已经结束，具有明确的起止时间点。

```python
class RangeStatusT(tp.NamedTuple):
    Open: int
    Closed: int

RangeStatus = RangeStatusT(*range(2))
```

例子：
```python
if range_record['status'] == RangeStatus.Open:
    print("范围仍在进行中")
elif range_record['status'] == RangeStatus.Closed:
    print("范围已结束")
```

# DrawdownStatus
用于定义回撤的状态：
- `Active`：回撤正在进行中，价值尚未恢复到峰值
- `Recovered`：回撤已经恢复，价值达到或超过了历史峰值

```python
class DrawdownStatusT(tp.NamedTuple):
    Active: int
    Recovered: int

DrawdownStatus = DrawdownStatusT(*range(2))
```
