# class SignalsAccessor(GenericAccessor)
信号访问器类 `SignalsSRAccessor` 和 `SignalsDFAccessor` 的基类，用于处理信号数据。
- 大量方法与[signals/nb.ipynb](./nb.ipynb)中方法有关

## `__init__`
参数
- `obj` (Series/DataFrame)：要处理的信号数据对象
  - 必须是 pandas 的 Series 或 DataFrame 类型
  - 数据类型必须是布尔型(np.bool_)
  - True 表示有信号，False 表示无信号
- `**kwargs`：传递给父类 GenericAccessor 的关键字参数

```python
def __init__(self, obj: tp.SeriesFrame, **kwargs) -> None:
    checks.assert_dtype(obj, np.bool_)
    GenericAccessor.__init__(self, obj, **kwargs)
```

# class SignalsSRAccessor(SignalsAccessor, GenericSRAccessor)
专门处理 `Series` 类型信号数据的信号访问器类。

注意：
```python
@register_series_vbt_accessor('signals')
class SignalsSRAccessor(SignalsAccessor, GenericSRAccessor): ...
```
参考[root_accessors.ipynb](../root_accessors.ipynb)：
- 对于一个 `Series` 对象 `obj`，显然 `obj.vbt.signals` 相当于 `SignalsSRAccessor(Vbt_SRAccessor(obj))`

## `__init__`
```python
def __init__(self, obj: tp.Series, **kwargs) -> None:
    GenericSRAccessor.__init__(self, obj, **kwargs)
    SignalsAccessor.__init__(self, obj, **kwargs)
```
