# 类继承关系
专门用于处理 *OHLCV（开盘价Open、最高价High、最低价Low、
收盘价Close、成交量Volume）金融数据* 的访问器模块
```mermaid
classDiagram
    class GenericDFAccessor
    class OHLCVDFAccessor
    
    %% 继承关系
    GenericDFAccessor <|-- OHLCVDFAccessor
```

# class OHLCVDFAccessor(GenericDFAccessor)
注意：
```python
@register_dataframe_vbt_accessor('ohlc')
@register_dataframe_vbt_accessor('ohlcv')
class OHLCVDFAccessor(GenericDFAccessor): ...
```
参考 [root_accessors.ipynb](./root_accessors.ipynb)，对于任意 `DataFrame` 的实例 `obj`
- `obj.vbt.ohlc/ohlcv` 相当于 `OHLCVDFAccessor(Vbt_DFAccessor(obj))`

## `__init__`
参数：
- `obj`：DataFrame 对象，包含 OHLCV 数据
- `column_names`：自定义列名映射字典，用于指定 OHLCV 列的实际名称
  - 例如
    ```python
    column_names = {
        'open': 'actual_open_column_name',
        'high': 'actual_high_column_name', 
        'low': 'actual_low_column_name',
        'close': 'actual_close_column_name',
        'volume': 'actual_volume_column_name'
    }
    ```
- `**kwargs`：传递给父类的其他配置参数

```python
def __init__(self, obj: tp.Frame, column_names: tp.KwargsLike = None, **kwargs) -> None:
    self._column_names = column_names
    GenericDFAccessor.__init__(self, obj, column_names=column_names, **kwargs)
```

### 例子

````python
import pandas as pd

df = pd.DataFrame({
    'open': [100, 105, 102, 108, 106],
    'high': [110, 108, 107, 112, 110], 
    'low': [98, 102, 100, 105, 104],
    'close': [105, 104, 106, 109, 107],
    'volume': [1000000, 1200000, 800000, 1500000, 1100000]
})

# 创建OHLCV访问器
ohlcv = df.vbt.ohlcv
print(ohlcv)

custom_ohlcv = df.vbt.ohlcv(column_names={
    'open': 'o', 'high': 'h', 'low': 'l', 'close': 'c', 'volume': 'v'
})
print(custom_ohlcv)
````

注意，例子中 `custom_ohlcv` 的 `df.vbt.ohlcv` 之后 `(column_names=...)` 实际调用的是 `BaseAccessor` 的 `__call__` 方法。
