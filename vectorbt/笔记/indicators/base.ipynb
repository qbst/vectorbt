# 类继承关系

```mermaid
classDiagram
    class MA
    class MSTD
    class BBANDS
    class RSI
    class STOCH
    class MACD
    class ATR
    class OB
    class _MA
    class _MSTD
    class _BBANDS
    class _RSI
    class _STOCH
    class _MACD
    class _ATR
    class _OB

    
    %% 继承关系
    MA <|-- _MA
    MSTD <|-- _MSTD
    BBANDS <|-- _BBANDS
    RSI <|-- _RSI
    STOCH <|-- _STOCH
    MACD <|-- _MACD
    ATR <|-- _ATR
    OB <|-- _OB
```
通过 [factory.py](../../indicators/factory.py) 中类 `IndicatorFactory` 的方法 `from_apply_func` 对 [nb.py](../../indicators/nb.py) 中定义的指标构建指标类型 `IndicatorBase`。

`_*` 类型就是在 `*` 类型的基础上增加了 `plot` 方法。
