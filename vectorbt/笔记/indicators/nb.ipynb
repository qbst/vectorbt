# 数学原理

## 移动平均
```python
@njit(cache=True)
def ma_cache_nb(close: tp.Array2d, windows: tp.List[int], ewms: tp.List[bool], adjust: bool) -> tp.Dict[int, tp.Array2d]
```

`ewm = False` 时为 **简单移动平均SMA**：
$$SM{A_t} = \left\{ \begin{array}{l}
NaN,t < w - 1\\
\frac{1}{w}\left( {{a_t} +  \cdots  + {a_{t - w + 1}}} \right),t \ge w - 1
\end{array} \right.$$
`ewm = True` 时为 **指数移动平均EMA**：
$$EM{A_t} = \left\{ \begin{array}{l}
{a_0},t = 0\\
\alpha_t  \cdot {a_t} + \left( {1 - \alpha_t } \right) \cdot EM{A_t},t > 0
\end{array} \right.$$
- `adjust = False` 时
$$\alpha_t  = \frac{2}{{w + 1}}$$
- `adjust = True` 时
$$\alpha_t  = \frac{1}{{1 +  \cdots  + {\beta ^t}}},\beta  = 1 - \frac{2}{w}$$

## 移动标准差
```python
@njit(cache=True)
def mstd_cache_nb(close: tp.Array2d, windows: tp.List[int], ewms: tp.List[bool], adjust: bool, ddof: int) -> tp.Dict[int, tp.Array2d]
```

令 $w = $ `window`，$d = $ `ddof` 

`ewm = False` 时为 **简单移动标准差**
$${\sigma _t} = \sqrt {\frac{{{S_t}}}{{w - d}}} $$
- 其中
$${\mu _t} = \frac{1}{w}\sum\limits_{t - w + 1 \le i \le t} {{a_i}} $$
$${S_t} = \sum\limits_{t - w + 1 \le i \le t} {{{\left( {{a_i} - {\mu _t}} \right)}^2}} $$
`ewm = True` 时为 **指数移动标准差**，令 $\alpha  = \frac{2}{{w + 1}}$
$${\sigma _t} = \sqrt {\frac{{{S_t}}}{{1 - {{\left( {1 - \alpha } \right)}^{t + 1}} - \delta_t }}} $$
- 其中
$${\mu _t} = \alpha  \cdot {a_t} + \left( {1 - \alpha } \right) \cdot {\mu _{t - 1}}$$
$${S_t} = \alpha  \cdot {\left( {{a_t} - {\mu _t}} \right)^2} + \left( {1 - \alpha } \right) \cdot {S_{t - 1}}$$
- `adjust = True` 时 $\delta_t  = 0$，否则 ${\delta _t} = d \cdot {\alpha ^2}{\left( {1 - \alpha } \right)^{2t}}$

## 布林带（Bollinger Bands）
```python
@njit(cache=True)
def bb_cache_nb(close: tp.Array2d, windows: tp.List[int], ewms: tp.List[bool], alphas: tp.List[float], adjust: bool, ddof: int) -> tp.Tuple[tp.Dict[int, tp.Array2d], tp.Dict[int, tp.Array2d]]
```

**布林带（Bollinger Bands）** 的定义：
- 中轨
$$Mid_t = 移动平均\left( {{a_t}} \right)$$
- 上轨
$$Upper_t = Mi{d_t} + \alpha  \cdot 移动标准差\left( {{a_t}} \right)$$
- 下轨
$$Lower_t = Mi{d_t} - \alpha  \cdot 移动标准差\left( {{a_t}} \right)$$

## 相对强弱指数（RSI）
```python
@njit(cache=True)
def rsi_cache_nb(close: tp.Array2d, windows: tp.List[int], ewms: tp.List[bool], adjust: bool) -> tp.Dict[int, tp.Tuple[tp.Array2d, tp.Array2d]]
```

首先定义
$$Gain_t = \max \left( {{a_t} - {a_{t - 1}},0} \right)$$
$$Loss_t = \max \left( { - \left( {{a_t} - {a_{t - 1}}} \right),0} \right)$$
**相对强度（RS）**
$$RS_t = \frac{{移动平均\left( Gain_t \right)}}{{移动平均\left( Loss_t \right)}}$$
**相对强弱指数（RSI）**
$$RSI_t = 100 - \frac{{100}}{{1 + RS_t}}$$

## 随机指标（Stochastic Oscillator）
```python
@njit(cache=True)
def stoch_cache_nb(high: tp.Array2d, low: tp.Array2d, close: tp.Array2d, k_windows: tp.List[int], d_windows: tp.List[int], d_ewms: tp.List[bool], adjust: bool) -> tp.Dict[int, tp.Tuple[tp.Array2d, tp.Array2d]]
```

定义
$${H_t} = \max \left( {high_{t - w + 1}}, \cdots ,high_t \right)$$
$${L_t} = \min \left( {low_{t - w + 1}}, \cdots ,low_t \right)$$
**K线**
$${K_t} = \frac{{clos{e_t} - {L_t}}}{{{H_t} - {L_t}}} \times 100\% $$
**D线**
$${D_t} = 移动平均\left( {{K_t}} \right) \times 100\% $$

## MACD（Moving Average Convergence Divergence）
```python
@njit(cache=True)
def macd_cache_nb(close: tp.Array2d, fast_windows: tp.List[int], slow_windows: tp.List[int], signal_windows: tp.List[int], macd_ewms: tp.List[bool], signal_ewms: tp.List[bool], adjust: bool) -> tp.Dict[int, tp.Array2d]
```

**MACD线**
$$MACD_t^{fast,slow} = 移动平均\left( {clos{e_t},{w_{fast}}} \right) - 移动平均\left( {clos{e_t},{w_{slow}}} \right)$$
**信号线**
$${信号线_t} = 移动平均\left( {MACD_t} \right)$$
**柱状图**
$${柱状图_t} = {MACD_t} - {信号线_t}$$

## 真实波幅（True Range）
```python
@njit(cache=True)
def true_range_nb(high: tp.Array2d, low: tp.Array2d, close: tp.Array2d) -> tp.Array2d
```

**真实波幅**
$$TR_t = \max \left( {hig{h_t} - lo{w_t},\left| {hig{h_t} - clos{e_{t - 1}}} \right|,\left| {lo{w_t} - clos{e_{t - 1}}} \right|} \right)$$

## 平均真实波幅（ATR）
```python
@njit(cache=True)
def atr_cache_nb(high: tp.Array2d, low: tp.Array2d, close: tp.Array2d, windows: tp.List[int], ewms: tp.List[bool], adjust: bool) -> tp.Tuple[tp.Array2d, tp.Dict[int, tp.Array2d]]
```

**平均真实波幅**
$${ATR_t} = 移动平均\left( {TR_t} \right)$$

## 能量潮指标（On-Balance Volume, OBV）

```python
@njit(cache=True)
def obv_custom_nb(close: tp.Array2d, volume_ts: tp.Array2d) -> tp.Array2d
```

定义价格变化方向矩阵
$$
D_t=\operatorname{sign}\left(\text { close }_t-\text { close }_{t-1}\right) \in\left\{\begin{array}{lll}
-1 & 0 & 1
\end{array}\right\}
$$
则 
$$OB{V_t} = \sum\limits_{i \le t} {{D_i} \cdot Volum{n_i}} $$
