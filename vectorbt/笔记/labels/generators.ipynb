# 类继承关系

```mermaid
classDiagram
    class FMEAN
    class FSTD
    class FMIN
    class FMAX
    class FIXLB
    class MEANLB
    class LEXLB
    class TRENDLB
    class BOLB

    
    %% 继承关系
    FIXLB <|-- _FIXLB
    MEANLB <|-- _MEANLB
    LEXLB <|-- _LEXLB
    TRENDLB <|-- _TRENDLB
    BOLB <|-- _BOLB
```

使用 [indicators/factory.py](../../indicators/factory.py) 中类 `IndicatorFactory` 的方法 `from_apply_func`，对 [labels/nb.py](../../indicators/nb.py) 中定义的标签指标构建指标类型 `IndicatorBase`。

`_*` 类型就是在 `*` 类型的基础上增加了 `plot` 方法。


例如
```python
def _plot(self, column: tp.Optional[tp.Label] = None, **kwargs) -> tp.BaseFigure:  # pragma: no cover
    self_col = self.select_one(column=column, group_by=False)
    return self_col.close.rename('close').vbt.overlay_with_heatmap(self_col.labels.rename('labels'), **kwargs)


class _FIXLB(FIXLB):
    plot = _plot
```
