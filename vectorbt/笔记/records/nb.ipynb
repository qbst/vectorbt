```python
import numpy as np
from vectorbt.records.nb import apply_on_mapped_nb, col_map_nb
from numba import njit

@njit
def standardize_nb(inout, idxs, col, values):
    mean_val = np.mean(values)
    std_val = np.std(values)
    if std_val > 0:
        inout[idxs] = (values - mean_val) / std_val
    else:
        inout[idxs] = np.zeros_like(values)

prices = np.array([100, 101, 99, 50, 51, 49])
stocks = np.array([0, 0, 0, 1, 1, 1])
col_map = col_map_nb(stocks, 2)
 
standardized = apply_on_mapped_nb(
    prices, col_map, standardize_nb
)
print(f"标准化后: {standardized}")
```
